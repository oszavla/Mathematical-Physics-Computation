```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

# Define the range of x values
x = np.linspace(0.001, 10, 200)

# Define the equations
y1 = 0.5*np.log(2*np.pi) + (x+0.5)*np.log(x)-x
y2 = x*np.log(x)-x
y3 = np.log(gamma(x+1))

# Plot the equations
plt.figure(figsize=(8, 6))
plt.plot(x, y1, label= r'$\frac{1}{2}ln(2\pi)+(n+\frac{1}{2})ln(n) - n $', linestyle='-', color='r')
plt.plot(x, y2, label=r'$n ln(n) - n $', linestyle='--', color='b')
plt.plot(x, y3, label=r'$ln(\Gamma(n+1))$', linestyle='-.', color='g')

# Add labels and title
plt.xlabel('x')
plt.ylabel('y')
plt.title('Approximation of Stirling Formula')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()


print(" x       |    1/2 ln(2π)+(n+1/2)ln(n) - n    |   nln(n) - n    |  ln[Γ(n+1)] ")
print("-" * 65)

for i in range(len(x)):
    print(f"{x[i]:>7.2f} | {y1[i]:>10.2f} | {y2[i]:>12.4f} | {y3[i]:>18.4f}")
```

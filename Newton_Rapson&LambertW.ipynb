```python
from scipy.optimize import newton
from scipy.special import lambertw
import numpy as np

def f(z, theta, m, k):
    u = -(k * z * np.sin(theta)) / m
    W = lambertw(u * np.exp(u)).real  # Ambil bagian real
    term1 = m * np.cos(theta) / k
    term2 = (m**3 * np.cos(theta) / np.sin(theta)**2 * np.exp((k * z * np.sin(theta)) / m) * (k**2 * z * np.sin(theta)**2 * np.exp(-(k * z * np.sin(theta)) / m) / m**2 - k * np.sin(theta) * np.exp(-(k * z * np.sin(theta)) / m) / m) * W) /(k**3 * z * (W + 1))
    return term1 - term2

theta = np.radians(30)  # Contoh sudut 30°
m, k = 1, 0.1
sol = newton(f, x0=10, args=(theta, m, k))
print(sol)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

# Define the range of x values
x = np.linspace(0.001, 20000, 200000)

# Define the equations
y1 = 0.5*np.log(2*np.pi) + (x+0.5)*np.log(x)-x
y2 = x*np.log(x)-x
y3 = np.log(gamma(x+1))

# Plot the equations
plt.figure(figsize=(8, 6))
plt.plot(x, y1, label= r'$\frac{1}{2}ln(2\pi)+(n+\frac{1}{2})ln(n) - n $', linestyle='-', color='r')
plt.plot(x, y2, label=r'$n ln(n) - n $', linestyle='--', color='b')
plt.plot(x, y3, label=r'$ln(\Gamma(n+1))$', linestyle='-.', color='g')

# Add labels and title
plt.xlabel('x')
plt.ylabel('y')
plt.title('Approximation of Stirling Formula')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

# Define the range of x values (excluding negative integers where Gamma is undefined)
x = np.linspace(-6, 6, 700)  # Range includes negative values (avoid -∞ at negative integers)
x = x[x != 0]  # Remove zero to prevent singularity

# Compute the Gamma function values
y = gamma(x+1)

# Plot the Gamma function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label=r'$\Gamma(x)$', color='b')

# Mark undefined points at negative integers
for i in range(-6, 1):
    plt.axvline(i, color='r', linestyle='dotted', alpha=0.6, label="Singularity" if i == -4 else "")

# Labels and grid
plt.xlabel("x")
plt.ylabel(r"$\Gamma(x)$")
plt.title("Plot of the Gamma Function")
plt.legend()
plt.ylim(-15, 15)  # Adjust y-axis for better visualization
plt.grid(True)

# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
from scipy.special import gamma

# Define the range of x values
x = np.linspace(100, 1500, 20000)

# Define the equations
y1 = 0.5*np.log(2*np.pi) + (x+0.5)*np.log(x)-x
y2 = x*np.log(x)-x
y3 = np.log(gamma(x+1))

# Specify the file path where you want to save the Excel file
excel_path = r"D:\Download\plot_data_stirling.xlsx"  # Change this to your desired location

# Save data to Excel with formatting
df = pd.DataFrame({'x': x, 'y1 (x^2)': y1, 'y2 (sin(x))': y2, 'y3 (exp(-0.1x) * cos(x))': y3})
with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='Data', index=False)
    workbook = writer.book
    worksheet = writer.sheets['Data']

    # Apply formatting
    from openpyxl.styles import Font, PatternFill
    header_font = Font(bold=True, color="FFFFFF")
    header_fill = PatternFill(start_color="4F81BD", end_color="4F81BD", fill_type="solid")

    for col in range(1, len(df.columns) + 1):
        cell = worksheet.cell(row=1, column=col)
        cell.font = header_font
        cell.fill = header_fill

print(f"Numerical values saved to '{excel_path}' with formatting.")

# Plot the equations
plt.figure(figsize=(8, 6))
plt.plot(x, y1, label= r'$\frac{1}{2}ln(2\pi)+(n+\frac{1}{2})ln(n) - n $', linestyle='-', color='r')
plt.plot(x, y2, label=r'$n ln(n) - n $', linestyle='--', color='b')
plt.plot(x, y3, label=r'$ln(\Gamma(n+1))$', linestyle='-.', color='g')


# Add labels and title
plt.xlabel('x')
plt.ylabel('y')
plt.title('Approximation of Stirling Formula')
plt.legend()
plt.grid(True)

# Save the plot as an SVG file
save_path = r"D:\Download"
os.makedirs(save_path, exist_ok=True)
file_path = os.path.join(save_path, "Stirling_Approximation.svg")
plt.savefig(file_path, format='svg')

print(f"Plot saved to '{save_path}' in SVG format.")


# Show the plot
plt.show()
```
